```python
#Install packages
import pandas as pd

#load data
air_data = pd.read_csv('AirQuality_Daily.csv')

#Make dataframe
air_data = pd.DataFrame(air_data)

print(air_data)
```

```python
# Select Relevant columns
air_data_selected = air_data[['sensor.name', 'pm2.5_atm', 'pm10.0_atm', 'voc']]

# Group Data by Sensor name and compute mean and median
summary_stats = air_data_selected.groupby('sensor.name').agg(
    pm25_mean=('pm2.5_atm', 'mean'),
    pm25_median=('pm2.5_atm', 'median'),
    pm10_mean=('pm10.0_atm', 'mean'),
    pm10_median=('pm10.0_atm', 'median'),
    voc_mean=('voc', 'mean'),
    voc_median=('voc', 'median'),
).reset_index()

# Round all numeric values to 2 decimal places
summary_stats = summary_stats.round(2)

# Identify top 5 locations by mean
top5_pm25_mean = summary_stats.sort_values(
    by='pm25_mean', ascending=False
).head(5)

top5_pm10_mean = summary_stats.sort_values(
    by='pm10_mean', ascending=False
).head(5)

top5_voc_mean = summary_stats.sort_values(
    by='voc_mean', ascending=False
).head(5)

# Identify top 5 locations by median
top5_pm25_median = summary_stats.sort_values(
    by='pm25_median', ascending=False
).head(5)

top5_pm10_median = summary_stats.sort_values(
    by='pm10_median', ascending=False
).head(5)

top5_voc_median = summary_stats.sort_values(
    by='voc_mean', ascending=False
).head(5)

# Display Results
print("Top 5 Locations by Mean PM2.5")
display(top5_pm25_mean)

print("Top 5 Locations by Median PM2.5")
display(top5_pm25_median)

print("Top 5 Locations by Mean PM10")
display(top5_pm10_mean)

print("Top 5 Locations by Median PM10")
display(top5_pm10_median)

print("Top 5 Locations by Mean VOC")
display(top5_voc_mean)

print("Top 5 Locations by Median VOC")
display(top5_voc_median)
```

```python
#Task 2

#Get max values of pollutants
pm25_max_value = air_data.agg(pm25_max=('pm2.5_atm','max')).iloc[0,0]
pm10_max_value = air_data.agg(pm10_max=('pm10.0_atm','max')).iloc[0,0]
voc_max_value = air_data.agg(voc_max=('voc','max')).iloc[0,0]

#Select the max rows of pollutant values
pm25_max_rows = air_data[air_data['pm2.5_atm'] == pm25_max_value]
pm10_max_rows = air_data[air_data['pm10.0_atm'] == pm10_max_value]
voc_max_rows = air_data[air_data['voc'] == voc_max_value]

#Make a summary in a table
voc_max_rows
```

```python
#Task 3
### Sample temperature function

# step 1: define function
def assign_temperature(temperature):
    if temperature < 32:
        return 'Below Freezing'
    elif 32 <= temperature <= 50:
        return 'Cool'
    elif 51 <= temperature <= 70:
        return 'Warm'
    else:
        return 'Hot'
#create column inside air data called temperature level
air_data['temperature_level'] = air_data['temperature'].apply(assign_temperature)


air_data.head(5)


#again with humidity
def assign_humidity(humidity):
    if humidity < 50:
        return 'Low'
    elif 50 <= humidity <= 80:
        return 'Medium'
    else:
        return 'High'
        
air_data['humidity_level'] = air_data['humidity'].apply(assign_humidity)


air_data.head(5)

###Make pivot tables with means
temperature_stats = pd.pivot_table(air_data,index='temperature_level', values=['pm2.5_atm','pm10.0_atm','voc'],aggfunc=['mean'])

temperature_stats
```

```python
### TASK 3 continued
#pivot table for humidity
humidity_stats = pd.pivot_table(air_data,index='humidity_level', values=['pm2.5_atm','pm10.0_atm','voc'],aggfunc=['mean'])

humidity_stats
```
